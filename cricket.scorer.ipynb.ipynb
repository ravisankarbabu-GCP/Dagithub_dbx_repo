```python
players = [

{'name': 'player1', 'runs': 0},
{'name': 'player2', 'runs': 0},
{'name': 'player3', 'runs': 0},
{'name': 'player4', 'runs': 0},
{'name': 'player5', 'runs': 0},
{'name': 'player6', 'runs': 0},
{'name': 'player7', 'runs': 0},
{'name': 'player8', 'runs': 0},
{'name': 'player9', 'runs': 0},
{'name': 'player10', 'runs': 0},
{'name': 'player11', 'runs': 0}
]

team_score = 0

striker = 0
non_striker = 1
next_player = 2

wickets = 0


while wickets < 10:

    event = input("Enter runs, W for wicket, Q for quit: ")

    if event.upper() == "Q":
        break

    if event.upper() == "W":

        wickets += 1

        if next_player < len(players):
            striker = next_player
            next_player += 1
    else:

        runs = int(event)

        players[striker]['runs'] += runs
        team_score += runs

        if runs % 2 == 1:
            temp = striker
            striker = non_striker
            non_striker = temp

    print("\n------------------")
    print("Team Score:", team_score)
    print("Wickets:", wickets)
    print("Current Striker:", players[striker]['name'])

    print("\nPlayer Scores")
    for player in players:
        print(player['name'], ":", player['runs'])
    print("------------------")
```
